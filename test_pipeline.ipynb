```python
import os
from dotenv import load_dotenv
# Notice the clean new import statement
from google import genai 

# 1. Load keys
load_dotenv()

# 2. Initialize the modern client
# It automatically looks for GOOGLE_API_KEY in your .env file
client = genai.Client()

try:
    print("Initializing modern Gemini 2.5 Flash pipeline...")
    
    # 3. Request a response using the correct 2026 model ID
    response = client.models.generate_content(
        model='gemini-2.5-flash',
        contents='Write one sentence about Macrina the Younger.',
    )
    
    print("\n--- Success! Modern Response ---")
    print(response.text)
    
except Exception as e:
    print("\n❌ Google Server Rejected the Call.")
    print(f"Error Details: {e}")
```
